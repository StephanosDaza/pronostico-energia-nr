```python
import pandas as pd
from energia import config, xm
```

```python
DESCARGAR = False

if DESCARGAR:
    xm.descargar_todo(config.FECHA_INICIO, config.FECHA_FIN)
```

```python
inventario = []
for codigo, (archivo, nombre, resolucion) in config.VARIABLES_XM.items():
    df = pd.read_csv(config.RAW / archivo)
    inventario.append({
        "metrica": codigo, "variable": nombre, "resolucion": resolucion,
        "filas": len(df), "columnas": df.shape[1],
        "desde": df["Date"].min(), "hasta": df["Date"].max(),
        "fechas_duplicadas": int(df["Date"].duplicated().sum()),
        "nulos": int(df.isna().sum().sum()),
    })
pd.DataFrame(inventario)
```
